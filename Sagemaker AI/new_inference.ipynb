```python
import sagemaker
from sagemaker.processing import ScriptProcessor, ProcessingInput, ProcessingOutput
from sagemaker import get_execution_role
import time

# ─── Setup ───────────────────────────────────────────────────────────────
region = "me-south-1"
session = sagemaker.Session()
role = get_execution_role()

timestamp = time.strftime("%Y-%m-%d-%H-%M-%S")
job_name = f"isolation-forest-batch-inference-hj-{timestamp}"
print(f"Starting batch inference job: {job_name}")

# ─── Paths ───────────────────────────────────────────────────────────────
model_s3_uri = (
    "s3://senseai-ml-results-bucket-hj20250426/model/training/"
    "train-isolation-forest-2025-06-02-09-34-20-224/output/model.tar.gz"
)

data_s3_uri = (
    "s3://senseai-ml-results-bucket-hj20250426/encoded/final_dataset/encoded_final.csv"
)

output_s3_uri = (
    f"s3://senseai-ml-results-bucket-hj20250426/inference-results/{job_name}/"
)

encoders_s3_uri = (
    "s3://senseai-ml-results-bucket-hj20250426/encoded/encoders/label_encoders.joblib"
)


processor = ScriptProcessor(
    image_uri="166555558375.dkr.ecr.me-south-1.amazonaws.com/shap-image",
    role=role,
    instance_count=1,
    instance_type="ml.r5.24xlarge",
    command=["python3"],
    base_job_name="isolation-forest-batch-inference",
    sagemaker_session=session
)

# ─── Run Job ─────────────────────────────────────────────────────────────
processor.run(
    code="inferencee.py",
    job_name=job_name,
    inputs=[
        ProcessingInput(source=model_s3_uri, destination="/opt/ml/processing/model"),
        ProcessingInput(source=data_s3_uri, destination="/opt/ml/processing/input"),
        ProcessingInput(source=encoders_s3_uri, destination="/opt/ml/processing/input/encoders")
    ],
    outputs=[
        ProcessingOutput(source="/opt/ml/processing/output", destination=output_s3_uri)
    ]
)

print(f"Job completed successfully. Output saved to:\n{output_s3_uri}")
```
