```python
import sagemaker
from sagemaker.sklearn.estimator import SKLearn
from sagemaker.inputs import TrainingInput
from datetime import datetime

# Initialize session and role
session = sagemaker.Session()
role = sagemaker.get_execution_role()

# Define S3 paths
input_s3_uri = "s3://senseai-ml-results-bucket-hj20250426/encoded/final_dataset/encoded_final.csv"
output_s3_uri = "s3://senseai-ml-results-bucket-hj20250426/model/training"

# Create SKLearn estimator
sklearn_estimator = SKLearn(
    entry_point="train_isolation_forest.py",
    role=role,
    instance_type="ml.m5.4xlarge",
    framework_version="1.2-1",
    py_version="py3",
    base_job_name="train-isolation-forest",
    sagemaker_session=session,
    output_path=output_s3_uri,
    code_location=output_s3_uri 
)

# Launch training job
sklearn_estimator.fit({
    "training": TrainingInput(s3_data=input_s3_uri, content_type="csv")
})
```
