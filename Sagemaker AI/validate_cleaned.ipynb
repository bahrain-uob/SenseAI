```python
from sagemaker.sklearn.processing import SKLearnProcessor
from sagemaker.processing import ProcessingInput, ProcessingOutput
from sagemaker import get_execution_role, Session

# ─── Setup ─────────────────────────────────────────────────────────
session = Session()
role = get_execution_role()

sklearn_processor = SKLearnProcessor(
    framework_version="1.2-1",
    role=role,
    instance_type="ml.m5.4xlarge",
    instance_count=1,
    base_job_name="validate-cleaned-final-updated",
    sagemaker_session=session
)

# ─── Run Validation Job ────────────────────────────────────────────
sklearn_processor.run(
    code="validate_preprocessed_cleaned.py",  # Your script above
    inputs=[
        ProcessingInput(
            source="s3://senseai-ml-results-bucket-hj20250426/cleaned/updated/preprocessed_cleaned_final_updated.csv",
            destination="/opt/ml/processing/input"
        )
    ],
    outputs=[
        ProcessingOutput(
            source="/opt/ml/processing/output",
            destination="s3://senseai-ml-results-bucket-hj20250426/validation-report-final-updated/"
        )
    ]
)
```
