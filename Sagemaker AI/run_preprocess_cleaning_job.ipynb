```python
from sagemaker.sklearn.processing import SKLearnProcessor
from sagemaker.processing import ProcessingInput, ProcessingOutput
import sagemaker
import boto3

# SageMaker setup
role = sagemaker.get_execution_role()
session = sagemaker.Session()
region = boto3.Session().region_name

# Processor
sklearn_processor = SKLearnProcessor(
    framework_version="1.2-1",
    role=role,
    instance_type="ml.m5.4xlarge",
    instance_count=1,
    base_job_name="preprocess-cleaned-final",
    sagemaker_session=session
)

# Run the processing job
sklearn_processor.run(
    code="preprocess_cleaned_final_ready.py",
    inputs=[
        ProcessingInput(
            source="s3://senseai-ml-results-bucket-hj20250426/cleaned/preprocessed_cleaned_ready.csv/",
            destination="/opt/ml/processing/input"
        )
    ],
    outputs=[
        ProcessingOutput(
            source="/opt/ml/processing/output",
            destination="s3://senseai-ml-results-bucket-hj20250426/cleaned/preprocessed_cleaned_final.csv/"
        )
    ]
)

```
