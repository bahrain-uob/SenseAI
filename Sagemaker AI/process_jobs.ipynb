```python
import sagemaker
from sagemaker.processing import ProcessingInput, ProcessingOutput, ScriptProcessor
from sagemaker.sklearn.processing import SKLearnProcessor 
from sagemaker import get_execution_role

# Define role and session
role = get_execution_role()
session = sagemaker.Session()
region = session.boto_region_name

# Define the processor (based on scikit-learn container)
processor = SKLearnProcessor(
    framework_version="1.2-1",  # Compatible version for scikit-learn
    role=role,
    instance_type="ml.m5.4xlarge",
    instance_count=1,
    base_job_name="decode-anomaly-scores",
    sagemaker_session=session,
)

# Define S3 locations
label_encoders_s3 = "s3://senseai-processed-transactions-dataset-hj20250426/processed/label_encoders.joblib"
anomaly_scores_s3 = "s3://senseai-ml-results-bucket-hj20250426/inference-results/batch-inference-hj-2025-05-02-08-09-35/anomaly_scores.csv"
output_s3 = "s3://senseai-ml-results-bucket-hj20250426/decoded-results/"

# Run processing job
processor.run(
    code="decode_anomaly_scores.py",
    inputs=[
        ProcessingInput(source=label_encoders_s3, destination="/opt/ml/processing/input/label_encoders"),
        ProcessingInput(source=anomaly_scores_s3, destination="/opt/ml/processing/input/anomaly_scores"),
    ],
    outputs=[
        ProcessingOutput(source="/opt/ml/processing/output", destination=output_s3),
    ],
)
```
